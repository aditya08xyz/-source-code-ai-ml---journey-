```python
# using groupby()
# so lets using the groupby , sumarise the data on monthly basis for better analysis
covid_month_df = covid_df.groupby('month')[['new_cases','new_deaths','new_tests']].sum()      #[['new_cases','new_deaths','new_tests']].sum()  to sum those on monthly basis
covid_month_df
# now we are getting a better view of our data
#similarly we can also check .mean() i.e avg of each month instead of .sum()
```

```python
#now
covid_weekday_df = covid_df.groupby('weekday')[['new_cases','new_deaths','new_tests']].mean()      # avg of each day 0=monday , 6=sunday 
covid_weekday_df
# so can see the most cases are found at sunday and most tests cinducted at friday , so maybe the test results takes 2 days to come and thats why sunday has most +ve cases , maybe
```

```python
covid_df['total_cases'] = covid_df.new_cases.cumsum()
covid_df['total_deaths'] = covid_df.new_deaths.cumsum()
covid_df['total_tests'] = covid_df.new_tests.cumsum() + initial_tests
covid_df
#cumsum genrally just adds value with each new case , and there are cummean(),cummax() etc...
```

```python
locations_df = pd.read_csv('location.csv')
locations_df
```

```python
locations_df[locations_df.location== "Italy"]
```

```python
# now since our above covid data was over the italy 
covid_df['location'] = 'Italy'      # its capital i in italy
covid_df
# covid_df['location'] = locations_df[locations_df.location]  , if it was the data from different different loaction
```

```python
# now to merge 
merge_df = covid_df.merge(locations_df, on ="location")   # on location , beacuse that was the common in both dataframe
merge_df
```

```python
merge_df['cases_per_million'] = merge_df.total_cases * 1e6 / merge_df.population
merge_df['deaths_per_million'] = merge_df.total_deaths * 1e6 / merge_df.population
merge_df['tests_per_million'] = merge_df.total_tests * 1e6 / merge_df.population
```

```python
merge_df
```

```python
# ahhh we ve done alot of analysis , and tons of columns 20 and much of it we dont need , so lets clean up
result_df = merge_df[['new_cases','total_cases','new_deaths','total_deaths','new_tests','total_tests','cases_per_million','deaths_per_million','tests_per_million']]
result_df
```

```python
# now to save the file will use .to_csv function
result_df.to_csv('results.csv' , index=None)           # index is none cuz i dont want it to make a cloumn of sno. i.e 1 to 247
# we made our analysis and put it in results.csv file can check in folder.
```
