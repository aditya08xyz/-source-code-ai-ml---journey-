```python
def parse_headers(header_line):
    return header_line.strip().split(',')       #it will take the value of any row of a csv file and remove /n and arrange them in a good manner like seprated with commas and all
```

```python
# eg
headers = parse_headers(file3_lines[0])  # here we are using the above function to create a perfect headers(we use file3_lines[0]) for us like the amount, rate, duration and all that ll help us later to arrange the data 
headers
```

```python
# to get the the float values that will assign those headers 
def parse_values(data_line):
    values=[]                                    # here data_line can be anything like file3_lines[2]
    for item in data_line.strip().split(','):
        if item == '':                            # we are using this function to take a clear value without/n and in float value so that if any thing like downpayment is zero , so the function must not fail , it say 0.0 as written in if statement, called handling missing values
            values.append(0.0)
        else:
            values.append(float(item))
    return values
```

```python
def create_item_dict(values, headers):
    result = {}
    for value, header in zip(values, headers):         # zip pairs the value alternate of each other like names= aditya,ujjwal no.=10,20   so if i use zip(name,no.)= (aditya,10),(ujjwal,20)
        result[header] = value     #important # since in dictionaries we cant append cuz it has key and value we cant append it like result.append(value,header) instead we use result[key] = value
    return result  
```

```python
def read_csv(path):
    result=[]
    #open the file in read mode
    with open(path, 'r') as f:
        #get the list of lines
        lines = f.readlines()
        # parse the headers
        headers = parse_headers(lines[0])
        
        for data_line in lines[1:]:
            #parse the values
            values = parse_values(data_line)
            # create a dictionary using values & headers
            item_dict = create_item_dict(values,headers)
            result.append(item_dict)
    return result
```

```python
def parse_headers(header_line):
    return header_line.strip().split(',')

def parse_values(data_lines):
    values= []
    for item in data_lines.strip().split(','):
        if item == '':
            values.append(0.0)
        else:
            values.append(float(item))
    return values

def create_item_dict(values, headers):
    result = {}
    for value,header in zip(values, headers):
        result[header]=value
    return result

def read_csv(path):
    result = []                          # {}is use for dictionaries and [] is use for lists and u can only append lists not dictionaries so we ve used [] .
    with open(path, 'r') as f:
        lines= f.readlines()
        headers = parse_headers(lines[0])
        for data_line in lines[1:]:
            values=parse_values(data_line)
            item_dict=create_item_dict(values, headers)
            result.append(item_dict)
    return(result)
```

```python
import math                 #now remember the previous notebook on calculating emi's , defining function on emi
def loan_emi(amount, duration, rate, down_payment=0):
    loan_amount= amount - down_payment
    try:
        emi = loan_amount * rate * (1+rate)**duration / ((1+rate)**duration - 1)
    except ZeroDivisionError:
        emi = loan_amount/duration
    emi = math.ceil(emi)
    return emi
```

```python
def write_csv(items, path):
    #open the file in write mode
    with open(path, 'w') as f:
        if len(items) == 0:  #return if theres nothing to write
            return
        
        #write a header in the first line                   #even i didnt understand this code , let focus on its working
        headers = list(items[0].keys())
        f.write(','.join(headers) + '\n')

        #write one item per line
        for item in items:
            values = []
            for header in headers:
                values.append(str(item.get(header,"")))
            f.write(','.join(values) + "\n")
```
