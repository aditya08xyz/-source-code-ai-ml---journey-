```python
def filter_even(number_list):
    result_list = []
    for number in number_list:                     # here number is a local variable ie it will only be used in this function and it will not be used outside the function in other cell meanwhile if we create a variable outside the function , it will be a global variable and can be used in every cell of this notebook.
        if number  % 2 == 0:
            result_list.append(number)
    return result_list
```

```python
#let suppose i want to buy a house worth 1.3 mil usd with 300k usd down payment and other with 8% interest rate per annum for 8 yr , so what would be my monthly emi  , emi = (P * r * (1+r)^n)/((1+r)^n - 1) where P=principal amount, r=interest rate, n=no of months

import math 

def loan_emi(amount, duration, rate, down_payment=0):
    loan_amount = amount - down_payment
    emi = loan_amount * rate * (1+rate)**duration / ((1+rate)**duration - 1)
    emi = math.ceil(emi)   #math.ceil is used to round off the value to the next integer value
    return emi
```

```python
#now we are going to modify our function of loan emi so that it can handle the the exception of rate =0
def updated_loan_emi (amount, duration, rate, down_payment=0):
    """calculates the equally monthly installment(emi) for a loan. """      #used to later check about ur function using help(updated_loan_emi)
    loan_amount = amount - down_payment # here loan amount is local variable can be used many times in differnt cell 
    try:
        emi = loan_amount * rate * (1 + rate)**duration / ((1+ rate)**duration - 1)
    except ZeroDivisionError:
        emi = loan_amount / duration
    emi = math.ceil(emi)
    return emi
```

```python
# for exmaple now i wanna calculate the amount of interest im paying monthly for a $100000 loan 10 yr 9%interest rate
emi_with_interest= updated_loan_emi(amount=100000, duration =10*12, rate=0.09/12 )
emi_with_interest
```

```python
emi_without_interest= updated_loan_emi(amount = 100000, duration  = 10*12, rate = 0)
emi_without_interest
```

```python
interest_amount_per_month= emi_with_interest - emi_without_interest
interest_amount_per_month
```

```python
total_interest= interest_amount_per_month * 10 *12
print("profit_of_bank:{}" .format(total_interest))
```
